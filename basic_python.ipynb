### Displaying a 3D Volume:

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate a 3D numpy array (e.g., 3D Gaussian blob)
x, y, z = np.meshgrid(np.linspace(-1, 1, 30),
                       np.linspace(-1, 1, 30),
                       np.linspace(-1, 1, 30))

vol = np.exp(-(x**2 + y**2 + z**2) * 10)  # 3D Gaussian shape




# Load a 3D volume
#vol = np.load('volume.npy')

# Display a slice of the volume using Matplotlib
plt.imshow(vol[:,:,15], cmap='gray')
plt.title('3D Volume Slice')
plt.show()
```

### Applying Thresholding to Medical Images:

```python
import cv2

# Load a medical image using OpenCV
img = cv2.imread('./data/sample_1.png', cv2.IMREAD_GRAYSCALE)

# Apply thresholding to the image using OpenCV
thresholded_img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)[1]

# Display the original and thresholded images using OpenCV
cv2_imshow( img)
cv2_imshow( thresholded_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Applying Edge Detection to Medical Images:

```python
import cv2

# Load a medical image using OpenCV
img = cv2.imread('./data/sample_3.png', cv2.IMREAD_GRAYSCALE)

# Apply edge detection to the image using OpenCV
edges = cv2.Canny(img, 100, 200)

# Display the original and edge-detected images using OpenCV
cv2_imshow( img)
cv2_imshow( edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Applying Morphological Operations to Medical Images:

```python
import cv2
import numpy as np

# Load a medical image using OpenCV
img = cv2.imread('./data/sample_1.png', cv2.IMREAD_GRAYSCALE)

# Define a structuring element for morphological operations
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))

# Apply morphological operations to the image using OpenCV
closed = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)
opened = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)

# Display the original and morphologically processed images using OpenCV
cv2_imshow( img)
cv2_imshow( closed)
cv2_imshow( opened)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Extracting and Displaying Contours from Medical Images:

```python
import cv2

# Load a medical image using OpenCV
img = cv2.imread('./data/sample_3.png', cv2.IMREAD_GRAYSCALE)

# Find the contours in the image using OpenCV
contours, hierarchy = cv2.findContours(img, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

# Draw the contours on a new image using OpenCV
contour_img = cv2.drawContours(cv2.cvtColor(img, cv2.COLOR_GRAY2BGR), contours, -1, (0, 0, 255), 2)

# Display the original and contour images using OpenCV
cv2_imshow( img)
cv2_imshow( contour_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Applying Filters to Medical Images:

```python
import cv2
import numpy as np

# Load a medical image using OpenCV
img = cv2.imread('./data/sample_1.png', cv2.IMREAD_GRAYSCALE)

# Apply a Gaussian filter to the image using OpenCV
filtered_img = cv2.GaussianBlur(img, (5, 5), 0)

# Display the original and filtered images using OpenCV
cv2_imshow( img)
cv2_imshow(filtered_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Performing Image Registration:

```python
import cv2

# Load two medical images using OpenCV
img1 = cv2.imread('./data/sample_1.png', cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread('./data/sample_3.png', cv2.IMREAD_GRAYSCALE)

# Detect and extract features from the images using OpenCV
detector = cv2.ORB_create()
kp1, des1 = detector.detectAndCompute(img1, None)
kp2, des2 = detector.detectAndCompute(img2, None)

# Match the features between the images using OpenCV
matcher = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
matches = matcher.match(des1, des2)

# Apply image registration using OpenCV
src_pts = np.float32([kp1[m.queryIdx].pt for m in matches]).reshape(-1, 1, 2)
dst_pts = np.float32([kp2[m.trainIdx].pt for m in matches]).reshape(-1, 1, 2)
M, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 5.0)

# Warp the second image to the first image's coordinate system using OpenCV
warped_img = cv2.warpPerspective(img2, M, (img1.shape[1], img1.shape[0]))

# Display the original and registered images using OpenCV
cv2_imshow( img1)
cv2_imshow( img2)
cv2_imshow(warped_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
